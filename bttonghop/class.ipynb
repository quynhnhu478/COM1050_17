**Bai 1** ```Một công ty bán lẻ cần phân tích các mã đơn hàng được tạo ra từ hệ thống bán hàng. Các mã đơn hàng này có định dạng không nhất quán, nhưng chứa các thông tin quan trọng như mã sản phẩm, số lượng và trạng thái của đơn hàng. Hãy viết một hàm Python duy nhất tên là parse_order_code(order_code) nhận vào một chuỗi order_code và trả về thông tin tóm tắt về mã sản phẩm, số lượng và trạng thái của đơn hàng.```
> Yêu cầu chi tiết:
- Chuẩn hóa mã đơn hàng: Chuyển toàn bộ order_code sang chữ thường (lowercase, cho phép sử dụng hàm .lower()).
- Trích xuất mã sản phẩm: Mã sản phẩm luôn có dạng prod_ trong mã đơn hàng. Nếu không tìm thấy, mặc định là "UNKNOWN".
- Trích xuất số lượng: Số lượng sản phẩm luôn xuất hiện sau chuỗi qty_ trong mã đơn hàng. Nếu không tìm thấy, mặc định là 1.
- Trích xuất trạng thái đơn hàng: Trạng thái đơn hàng có thể là "pending", "shipped", "delivered" hoặc "cancelled".
- Nếu không tìm thấy trạng thái, mặc định là "pending".
- Nếu có nhiều trạng thái xuất hiện trong chuỗi, trả về trạng thái đầu tiên tìm thấy.

> Kết quả trả về (Output): Hàm phải trả về một chuỗi tóm tắt theo cấu trúc: "Product: [Mã sản phẩm]. Quantity: [Số lượng]. Status: [Trạng thái]."

> input: ORDER_PROD_ABC123_QTY_5_STATUS_SHIPPED
> output: 
- Product: abc123.
- Quantity: 5.
- Status: shipped. 


````python
def parse_order_code(order_code):
    code = order_code.lower() # chuyển toàn bộ sang chữ thường
    product = "UNKNOWN" 
    if "prod_" in code: #nếu xuất hiện prod_ trong code
        part = code.split("prod_")[1] #split tách prod_ thành list 2 phần: phần trước prod_ có vị trí 0, còn phần sau vị trí 1
        product = part.split("_")[0] #mỗi phần tách bởi dấu _ sẽ đóng vị trí bắt đầu từ 0
    quantity = "1" #luôn gắn qty = 1 nếu không có qty trong code
    if "qty_" in code: 
        part = code.split("qty_")[1] 
        quantity = part.split("_")[0] #phần tử trước _ là số lượng
    status = "pending"#status luôn gắn=pending
    statuses = ["pending", "shipped", "delivered", "cancelled"]
    first = len(code) #lấy ptu xuat hien dau của status
    for i in statuses:
        vt = code.find(i) #tìm trạng thái 
        if vt != -1 and vt < first:#vị trí đầu tiên xuất hiện status 
            first = vt #cập nhật lại vị trí để tìm xuất hiện sớm nhất
            status = i
 
    return f"Product: {product}. \nQuantity: {quantity}. \nStatus: {status}."
order_code=input().strip()
print(parse_order_code(order_code))
````

**Bài 3** Tính diện tích toàn phần và thể tích hình trụ với pi=3.14

````python
class Cylinder:
    def __init__(self, radius, height):
        self.x=radius
        self.y=height
    def surface_area(self):
        return 2*3.14*self.x*self.y+2*3.14*(self.x**2)
    def volume(self):
        return 3.14*(self.x**2)*self.y
r,h=map(float,input().split())
cylinder=Cylinder(r,h)
area=cylinder.surface_area()
vol=cylinder.volume()
print(f"{area:.2f} {vol:.2f}")
````

**Bài 5** Kiểm tra xem x,y có phải là gốc tọa độ không, có nằm trên trục tung, trục hoành không, tính khoảng cách từ điểm đó đến gốc tọa độ, lấy 2 số phần thập phân 
```
input: 
3 4 
> M khong nam tren truc nao 
  5.00

3 0 
> M nam tren truc hoanh
  3.00

0 -9 
> M nam tren truc tung
  9.00

7.8 4.5
> M khong nam tren truc nao 
  9.00

````python
import math
class Point:
    def __init__(self, x, y):
        self.x=x
        self.y=y
    def is_origin(self):
        if self.x==0 and self.y==0:
            return True
        return False
    def is_on_y_axis(self):
        if self.x==0 and self.y!=0:
            return True
        return False
    def is_on_x_axis(self):
        if self.x!=0 and self.y==0:
            return True
        return False
    def distance_to(self,other):
        return math.sqrt((self.x-other.x)**2+(self.y-other.y)**2)
x,y=map(float,input().split())
M=Point(x,y)
O=Point(0,0)
if M.is_origin():
    print("M la goc toa do")
elif M.is_on_y_axis():
    print("M nam tren truc tung")
elif M.is_on_x_axis():
    print("M nam tren truc hoanh")
else:
    print("M khong nam tren truc nao")
dist=M.distance_to(O)
print(f"{dist:.2f}")
````

```

````python
class BankAccount:
    def __init__(self,owner,balance):
        self.owner=owner
        self.balance=float(balance)
    def deposit(self,amount):
        self.balance+=amount
    def withdraw(self,amount):
        if self.balance>=amount:
            self.balance-=amount
    def __str__(self):
        print(f"{self.balance:.2f}")
s,b=input().split()
n=int(input())
A=BankAccount(s,b)
for i in range(n):
    x,y=input().split()
    y=float(y)
    if x=="DEPOSIT":
        A.deposit(y)
    elif x=="WITHDRAW":
        A.withdraw(y)
A.__str__()
````

**Bài 7** Cộng trừ nhân chia với phân số theo dạng a,b,op,c,d
với:
> a,b là tử, mẫu pso 1

> op là dấu 
> c,d là tử, mẫu pso 2
```
in ra kết quả tối giản. Dấu ở mẫu thì phải chuyển lên tử hết
input: 2 4 - 1 4  -> 1/4
-2 3 * 3 5  -> -2/5
7 8 / 14 3  -> 3/16
1 2 +  1 3  -> 5/6

````python
import math
class Fraction:
    def __init__(self, num, den):
        if den<0:
            self.tu=-num//(math.gcd(num,den))
            self.mau=-den//(math.gcd(num,den))
        self.tu=num//(math.gcd(num,den))
        self.mau=den//(math.gcd(num,den))
    def add(self,other):
        t=math.lcm(self.mau,other.mau)
        self.tu=self.tu*t//self.mau+(other.tu*t)//other.mau
        self.mau=math.lcm(self.mau,other.mau)
    def sub(self,other):
        t=math.lcm(self.mau,other.mau)
        self.tu=self.tu*t//self.mau-(other.tu*t)//other.mau
        self.mau=math.lcm(self.mau,other.mau)
    def mul(self,other):
        self.tu*=other.tu
        self.mau*=other.mau
    def div(self,other):
        self.tu*=other.mau
        self.mau*=other.tu
    def toigian(self):
        gcd=math.gcd(self.tu,self.mau)
        self.tu=self.tu//gcd
        self.mau=self.mau//gcd
    def __str__(self):
        self.toigian()
        print(f"{self.tu}/{self.mau}")
a,b,op,c,d=input().split()
c=int(c)
d=int(d)
A=Fraction(int(a),int(b))
B=Fraction(c,d)
if op=="+":
    A.add(B)
elif op=="-":
    A.sub(B)
elif op=="*":
    A.mul(B)
else:
    A.div(B)
A.__str__()
````

**Bài 8** cho 4 điểm A B C và D trong hệ tọa độ 3 chiều 
    in ra góc tạo bởi mp(A,B,C) và (B,C,D)
```
 input:
 0 4 5 
 1 7 6 
 0 5 9 
 1 7 2 
 output:8.19
 


````python
import math

class Point:
    def __init__(self, x, y, z):
        self.x, self.y, self.z = x, y, z

    # Nạp chồng toán tử trừ để tạo Vecto từ 2 điểm
    def __sub__(self, other):
        return Point(self.x - other.x, self.y - other.y, self.z - other.z)

    # Tích có hướng (Cross Product) để tìm pháp tuyến
    def cross(self, other):
        return Point(self.y*other.z - self.z*other.y,
                     self.z*other.x - self.x*other.z,
                     self.x*other.y - self.y*other.x)

    # Tích vô hướng (Dot Product)
    def dot(self, other):
        return self.x*other.x + self.y*other.y + self.z*other.z

    # Độ dài vecto
    def absolute(self):
        return math.sqrt(self.x**2 + self.y**2 + self.z**2)

# Nhập 4 điểm
points = []
for _ in range(4):
    points.append(Point(*map(float, input().split())))

a, b, c, d = points

# Vecto pháp tuyến
n1 = (b - a).cross(c - b)
n2 = (c - b).cross(d - c)

# Tính cos PHI
cos_phi = n1.dot(n2) / (n1.absolute() * n2.absolute())
# Đảm bảo giá trị nằm trong khoảng [-1, 1] để tránh lỗi toán học
cos_phi = max(-1, min(1, cos_phi))

phi_rad = math.acos(cos_phi)
phi_deg = math.degrees(phi_rad)

print(f"{phi_deg:.2f}")
````
