**Bai 1** ```Tính tổng k liên tiếp lớn nhất```
> input:
- 6 3
- 1 3 -2 5 4 -1
> output:
- 8

````python
#sử dụng mảng cộng dồn
def max_subarray_sum(a,k):
    n=len(a)
    gmax=float('-inf')
    s=[0]*(n+1)
    n=len(a)
    for i in range(1,n+1):
     s[i]=s[i-1]+a[i-1]
    for i in range(n):
     for j in range(i+1,n+1):
       if (s[j]-s[i]>gmax) and (j-i)==k:
        gmax=s[j]-s[i]
    return(gmax)
n,k=map(int,input().split())
a=list(map(int,input().strip().split()))
print(max_subarray_sum(a,k))
````

> input: helloo4hG

> output: Valid

````python
def is_valid_password(p):
    kt=False
    chuthuong=False
    chuhoa=False        
    number=False
    d=len(p) #length 
    #c2 tính length: l=p.count("")-1 
    for i in p:
        if i.islower():
            chuthuong=True
        if i.upper():
            chuhoa=True
        if i.isdigit():
            number=True
    if d>=8 and chuthuong and chuhoa and number:
        return("Valid")
    else:
        return("Invalid") 
password=input().strip()
print(is_valid_password(password))

    
````
