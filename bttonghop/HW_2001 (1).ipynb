## Câu 1. Chuỗi – thống kê từ

Ví dụ:


## Câu 2. Dãy con không giảm

Ví dụ:
```
Input:
1 3 2 2 4

Output:
4
```

````python
n=int(input())
a=list(map(int,input().strip().split()))
dp=[1]*(n)
for i in range(n):
    for j in range(i):
        if a[i]>a[j]:
            dp[i]=max(dp[i],dp[j]+1)
print(max(dp))
````

## Câu 4. Ký tự xuất hiện nhiều nhất

Cho một chuỗi.
Tìm ký tự xuất hiện nhiều nhất. Nếu có nhiều hơn 1 ký tự có số lần xuất hiện giống nhau, lấy ký tự nhỏ nhất theo thứ tự từ điển.

Ví dụ:
```
Input:
banana

Output:
a
```

````python
s = input()
max_char = ""
max_count = 0
for char in sorted(set(s)): 
    print(char)
    count = s.count(char) # Đếm số lần char xuất hiện trong s
    if count > max_count:
        max_count = count
        max_char = char
print(max_char)
````

## Câu 5. Chia dãy thành hai phần

Cho dãy số.
Kiểm tra xem có thể chia dãy thành hai phần có tổng bằng nhau hay không.

Ví dụ:
```
Input:
1 5 11 5

Output:
YES
```

````python
nums = [1, 5, 11, 5]
total = sum(nums)
if total % 2 != 0:
    print("NO")
else:
    n = total // 2
    check = [False] * (n + 1)
    check[0] = True # Tổng bằng 0 luôn đúng
    for x in nums:
        # Duyệt ngược từ target về x để tránh cộng lặp một số nhiều lần
        for j in range(n, x - 1, -1):
            if check[j - x] == True:
                check[j] = True
                
    if check[n]:
        print("YES")
    else:
        print("NO")
````

## Câu 6. Ma trận – đường chéo

Cho ma trận n x n.
Tìm phần tử lớn nhất trên mỗi đường chéo chính và phụ.

Ví dụ:
```
Input:
3
1 2 3
4 5 6
7 8 9

Output:
9 7
```

````python
# Bước 1: Nhập kích thước ma trận
n = int(input("Nhập n: "))

# Bước 2: Nhập ma trận (từng hàng một)
matrix = []
for i in range(n):
    # Nhập một dòng, tách các số bằng khoảng trắng và chuyển thành số nguyên
    row = list(map(int, input(f"Nhập hàng {i+1}: ").split()))
    matrix.append(row)

# Bước 3: Khởi tạo giá trị lớn nhất (lấy ngay phần tử đầu của mỗi đường chéo)
max_main = matrix[0][0]
max_sub = matrix[0][n-1]

# Bước 4: Duyệt qua các đường chéo bằng 1 vòng lặp duy nhất
for i in range(n):
    # Đường chéo chính: hàng i, cột i
    if matrix[i][i] > max_main:
        max_main = matrix[i][i]
    
    # Đường chéo phụ: hàng i, cột n-1-i
    if matrix[i][n-1-i] > max_sub:
        max_sub = matrix[i][n-1-i]

# Bước 5: In kết quả
print(max_main, max_sub)
````

## Câu 7. Ma trận nhị phân – miền liên thông

Trong ma trận nhị phân, mỗi ô có giá trị:

1 : ô có dữ liệu (ví dụ: đất, vật thể, vùng cần quan tâm)

0 : ô trống / không xét

Miền liên thông các số 1 là một tập các ô có giá trị 1, sao cho:

Từ bất kỳ ô nào trong tập, ta đi được tới các ô còn lại

Việc “đi” chỉ được phép theo 4 hướng: Trên (up), Dưới (down), Trái (left), Phải (right)

Không được đi chéo

Với đầu vào là một ma trận nhị phân.
Đếm số miền liên thông các số 1 (4 hướng).



Ví dụ:
```
Input:
3 3
1 1 0
0 1 0
1 0 1

Output:
3
```

````python
import sys
# Tăng giới hạn đệ quy để tránh lỗi với ma trận lớn
sys.setrecursionlimit(2000)

def giai_quyet():
    # Nhập kích thước R x C
    try:
        r, c = map(int, input().split())
    except EOFError:
        return

    grid = []
    for _ in range(r):
        grid.append(list(map(int, input().split())))

    def dfs(i, j):
        # Nếu ra ngoài biên hoặc gặp ô trống (0) thì dừng
        if i < 0 or i >= r or j < 0 or j >= c or grid[i][j] == 0:
            return
        
        # Đánh dấu ô này đã xét bằng cách biến nó thành 0
        grid[i][j] = 0
        
        # Loang ra 4 hướng: Trên, Dưới, Trái, Phải
        dfs(i + 1, j)
        dfs(i - 1, j)
        dfs(i, j + 1)
        dfs(i, j - 1)

    count = 0
    for i in range(r):
        for j in range(c):
            if grid[i][j] == 1:
                count += 1
                dfs(i, j) # Loang để xóa toàn bộ miền này
    
    print(count)

giai_quyet()
````

## Câu 10. File – 5 số lớn nhất

Cho file chứa các số nguyên.

Chạy block ở  dưới, sau đó thực hiện tìm 5 số lớn nhất không trùng nhau trong file.

Ví dụ:
```
Input:
data.txt

Output:
100 98 97 90 88
```

````python
# Giả sử file của bạn tên là 'data.txt'
file_name = 'data.txt'

try:
    with open(file_name, 'r') as f:
        # 1. Đọc tất cả các số và chuyển về kiểu integer
        # split() giúp tách các số dựa trên khoảng trắng hoặc xuống dòng
        numbers = [int(x) for x in f.read().split()]

    # 2. Loại bỏ các số trùng nhau bằng set()
    unique_numbers = list(set(numbers))

    # 3. Sắp xếp giảm dần
    unique_numbers.sort(reverse=True)

    # 4. Lấy 5 phần tử đầu tiên
    result = unique_numbers[:5]

    # In kết quả theo định dạng yêu cầu
    print(*(result))

except FileNotFoundError:
    print("Không tìm thấy file data.txt")
except Exception as e:
    print(f"Có lỗi xảy ra: {e}")
````

Cho file sinh viên.

````python
with open("students.txt", "w") as f:
    names = ["An", "Binh", "Chi", "Dung", "Hieu", "Khanh", "Long", "Minh", "Nam", "Phuong"]
    for _ in range(random.randint(8, 15)):
        name = " ".join(random.sample(names, random.randint(1, 2)))
        score = round(random.uniform(4, 10), 1)
        f.write(f"{name} {score}\n")
````

## Câu 20. Bài toán tổng hợp

Quản lý chi tiêu cá nhân từ file.
Tính tổng thu, tổng chi và ngày chi tiêu nhiều nhất.

Ví dụ:
```
Input:
transactions.txt

Output:
Báo cáo thống kê
```

````python
with open("transactions.txt", "w") as f:
    days = [f"{i:02d}" for i in range(1, 29)]
    for _ in range(random.randint(10, 30)):
        day = random.choice(days)
        typ = random.choice(["IN", "OUT"])
        amount = random.randint(10, 500)
        f.write(f"{day} {typ} {amount}\n")
````
