**Bai 2** ```Ptu xuat hien nhieu nhat va so lan xuat hien```

````python
def xuathiennhieunhat(a):
    maxx = 0
    result = a[0]
    for i in a:
        dem = a.count(i)
        if dem > maxx:
            maxx = dem
            result = i
    return (f"{result} xuat hien nhieu nhat, som nhat, {maxx} lan")
a=list(map(int,input().split()))
print(xuathiennhieunhat(a)) 
````

**Bai 3** ```Đếm các cặp a,b sao cho a+b=X```

````python
def demcap(a,x):
    dem = 0
    for i in range(len(a)):
        for j in range(i+1,len(a)):
            if a[i] + a[j] == x:
                dem += 1
    return dem
a=list(map(int,input().split()))
x=int(input())
print(demcap(a,x))    
````

**Bai 5** ```Bạn được cung cấp: 1 dsach các khoảng thời gian interval=[[l1,r1],[l2,r2],..] và 1 danh sách các truy vấn queries = [q1,q2,...]. Với mỗi truy vấn q, hãy tìm độ dài nhỏ nhất của khoảng thời gian [li,ri] sao cho li<=q<=ri, nếu ko tồn tại in ra -1.```
```Chú ý: Cố giải quyết bài toán với độ phức tạp < O(m*n), hint: binary search in the queries list```
- Ví dụ: [[1,4],[2,4],[3,6],[4,4]] [2,3,4,5]-> [3,3,1,4]

````python
def nhonhat(a, q):
    b = []
    for i in q:
        minn = float('inf')  # Khởi tạo độ dài nhỏ nhất là vô cùng
        kt = False
        for l, r in a:
            if l <= i <= r:
                dodai = r - l + 1
                if dodai < minn:
                    minn = dodai
                kt = True
        if kt:
            b.append(minn)
        else:
            b.append(-1)
    return b
a = eval(input())
q = eval(input())
print(nhonhat(a,q))
````

**Bai 6** ```Tim do dai day con tang nghiem ngat dai nhat(khong nhat thiet phai lien ke)```
- Ví dụ: [10,9,2,5,3,7,101,18] -> 4 (2,3,7,10 hoac 2,3,7,18)

````python
def daycontangdainhat(a):
    if not a:#nếu a rỗng 
        return 0
    # dp[i] là độ dài dãy con tăng dài nhất kết thúc tại vị trí i
    dp = [1] * len(a)  # Mỗi phần tử ít nhất có độ dài 1 -> phần tử [1]*độ dài của dãy ->[1,1,...1]
    for i in range(1, len(a)):  
        for j in range(i):  
            if a[j] < a[i]:  
                dp[i] = max(dp[i], dp[j] + 1)  
    return max(dp)  
a = eval(input())
print(daycontangdainhat(a))
````

**Bai 8** ```Với bất kỳ mảng con nums[i:j] nào, hãy tính sums(num[i:j])%m và trả về kết quả lớn nhất có thể. Gợi ý: sử dụng tìm kiếm nhị phân để tìm tổng tiền tố nhỏ nhất chỉ lớn hơn tổng hiện tại một chút```
- vd: 


````python
def tongmax(a, m):
    n = len(a)
    # Mảng cộng dồn
    s = [0] * (n + 1)
    for i in range(1, n + 1):
        s[i] = s[i - 1] + a[i - 1]
    maxx = 0
    for i in range(n):
        for j in range(i + 1, n + 1):
            t = s[j] - s[i] 
            sums = t % m
            maxx = max(maxx, sums)
    return maxx
nums = eval(input())
m = int(input())
print(tongmax(nums, m))  
````

**Bai 10** ```In ra độ dài dãy con mà gtln và gtnn của dãy các nhau đúng bằng 1```
- vd: 1 3 2 2 5 2 3 7 -> [3,2,2,2,3]


````python
def dayhaihoa(a):
    n = len(a)
    d = 0
    for i in range(n):
        minn = a[i]
        maxx = a[i]
        for j in range(i, n):
            if a[j] < minn:
                minn = a[j]
            if a[j] > maxx:
                maxx = a[j]
            if maxx - minn == 1:
                s = j - i + 1
                if s > d:
                    d = s
                    vt1 = i
                    vt2 = j
            elif maxx - minn > 1:
                break   

    return vt1, vt2
arr = eval(input())
d, vt1, vt2 = dayhaihoa(arr)
# In kết quả
if d > 0:
    print(arr[vt1:vt2+1])
else:
    print("[]")
````

**Bai 13** ```Viết hàm tìm kiếm nhị phân 1 dsach đã được sxep tăng dần. Trả về chỉ số của ptu hoặc -1 nếu k tìm thấy ```
-vd: 


````python
def binary_search(a, x):
    left = 0
    right = len(a) - 1
    while left <= right:
        mid = (left + right) // 2
        if a[mid] == x:
            return mid
        elif a[mid] < x:
            left = mid + 1
        else:
            right = mid - 1

    return -1
ds = eval(input())
x=int(input())
print(binary_search(ds, x))  
````

**Bai 14** ```Cho danh sách a và 1 số nguyên x, đếm xem x xuất hiện bao nhiêu lần trong danh sách. Xây dựng hàm def count_occcurrences(arr:list, x:int) ->int trả về số lần xuất hiện của x trong arr ```
-vd: 


````python
def count_occcurrences(arr: list, x: int):
    count = 0
    for i in arr:
        if i == x:
            count += 1
    return count
arr = eval(input())
x = int(input())
print(count_occcurrences(arr, x))
````
