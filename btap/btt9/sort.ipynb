**BÀI 2** ```Viết hàm thực hiện thuật toán bubble sort hoặc thuật toán Selection Sort để sắp xếp danh sách theo thứ tự tăng dần. Sắp xếp nổi bọt (Bubble Sort); Sắp xếp chọn (Selection Sort)```

````python
def bubble_sort(a):
  n=len(a)
  for i in range(n):
    swapped=False
    for j in range(0,n-i-1):
      if a[j]>a[j+1]:
        a[j],a[j+1]=a[j+1],a[j]
        swapped=True
    if not swapped:
      break
  return(a)
a=list(map(int,input().strip().split()))
print(*bubble_sort(a))
````

**BÀI 4** ```Sắp xếp có điều kiện: Cho một dãy số nguyên. Bạn hãy giúp sắp xếp lại dãy số này sao cho số chẵn đứng trước, số lẻ đứng sau, chẵn tăng dần, lẻ giảm dần```
- VD:
- input output
- [1, 3, 2, 4, 6, 7, 23, 12] [2, 4, 6, 12, 23, 7, 3, 1]

````python
def sapxep(a):
  c=[]
  l=[]
  n=len(a)
  for i in range(n):
    if a[i]%2==0:
      c.append(a[i])
    else:
       l.append(a[i])
  return(sorted(c)+sorted(l,reverse=True))
a=eval(input())
print(f" {sapxep(a)}")
````
