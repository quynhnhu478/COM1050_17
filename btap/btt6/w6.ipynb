**W6A2** ```in ra tổng tích lũy (phần tử i là tổng từ đầu tới vị trí i), giữ nguyên thứ tự```

````python
a=list(map(int,input().split()))
n=len(a)
s=[0]*n
for i in range(0,n):
    s[i]=s[i-1]+a[i]
    print(s[i],end=' ')
````
